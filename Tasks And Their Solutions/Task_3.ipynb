**1. use the apply() function to categorize individuals as "High Income" or "Low Income" based on whether their AnnualIncome is above or below the median income?**

**2. How can you apply a function to calculate the difference between InterestRate and BaseInterestRate for each row and add it as a new column called RateDifference? ​​**







```python
import pandas as pd
```

```python
loan_df = pd.read_csv("/content/Loan.csv")
```

**use the apply() function to categorize individuals as "High Income" or "Low Income" based on whether their AnnualIncome is above or below the median income?**

```python

median_income = loan_df['AnnualIncome'].median()


def categorize_income(income):
    if income > median_income:
        return "High Income"
    else:
        return "Low Income"

loan_df['IncomeCategory'] = loan_df['AnnualIncome'].apply(categorize_income)


loan_df[['AnnualIncome', 'IncomeCategory']].head()

```

```python
#SAME WITH LAMBDA FUNCTION


loan_df['IncomeCategory'] = loan_df['AnnualIncome'].apply(lambda x: 'High Income' if x > median_income else 'Low Income')

loan_df[['AnnualIncome', 'IncomeCategory']].head()
```

**How can you apply a function to calculate the difference between InterestRate and BaseInterestRate for each row and add it as a new column called RateDifference? ​​**

```python
def calculate_rate_difference(row):
    return row['InterestRate'] - row['BaseInterestRate']

loan_df['RateDifference'] = loan_df.apply(calculate_rate_difference, axis=1)
loan_df[['InterestRate', 'BaseInterestRate', 'RateDifference']].head()
```

```python
#SAME WITH LAMBDA FUNCTION
loan_df['RateDifference'] = loan_df.apply(lambda row: row['InterestRate'] - row['BaseInterestRate'], axis=1)

loan_df[['InterestRate', 'BaseInterestRate', 'RateDifference']].head()
```
