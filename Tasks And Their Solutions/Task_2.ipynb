***Pivot Table***

**1. create a pivot table that shows the average LoanAmount for different EmploymentStatus and EducationLevel groups?**

**2. combination of MaritalStatus and RiskScore? How can you build a pivot table
to display the total MonthlyLoanPayment for each**

```python
import pandas as pd
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
#ignore this head part
loan_df.head()
```

**How can you create a pivot table that shows the average LoanAmount for different EmploymentStatus and EducationLevel groups?**

```python
pivot_1 = loan_df.pivot_table(values='LoanAmount', index='EmploymentStatus', columns='EducationLevel', aggfunc='mean')
pivot_1
```

**combination of MaritalStatus and RiskScore? How can you build a pivot table
to display the total MonthlyLoanPayment for each**

```python
loan_df.pivot_table(values='MonthlyLoanPayment', index='HomeOwnershipStatus', columns='RiskScore', aggfunc='sum')
```
