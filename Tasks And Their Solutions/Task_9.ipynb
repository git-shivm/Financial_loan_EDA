***Age and Anual income distribution with hist and KDE***

**1 . What is the distribution of borrower ages in the loan dataset, and how does it look when visualized using a histogram with a KDE (Kernel Density Estimate) overlay?**



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
#some data clean up
loan_df['ApplicationDate'] = pd.to_datetime(loan_df['ApplicationDate'])
loan_df.info()
```

```python
# Age distribution
plt.figure(figsize=(8, 6))
sns.histplot(loan_df['Age'], bins=30, kde=True, color='skyblue')
plt.title('Age Distribution', fontsize=16)
plt.show()

```

**2. What is the distribution of borrower Annual income in the loan dataset, and how does it look when visualized using a histogram with a KDE (Kernel Density Estimate) overlay?**

```python
# Annual Income Distribution
plt.figure(figsize=(8, 6))
sns.histplot(loan_df['AnnualIncome'], bins=50, kde=True, color='orange')
plt.title('Annual Income Distribution', fontsize=16)
plt.xlabel('Annual Income(USD)')
plt.ylabel('Frequency')
plt.show()
```
