***Loan Duration and Interest Rate :***

**Create a line plot showing the trend of average InterestRate over increasing LoanDuration. Add error bars to indicate the standard deviation at each point.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
#
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
```

```python


# Group by LoanDuration and calculate the mean and standard deviation of InterestRate
loan_duration_stats = loan_df.groupby('LoanDuration')['InterestRate'].agg(['mean', 'std']).reset_index()

# Set up the plot
plt.figure(figsize=(10, 6))
sns.set(style="whitegrid")

# Plot the line with error bars (standard deviation)
plt.errorbar(loan_duration_stats['LoanDuration'], loan_duration_stats['mean'],
             yerr=loan_duration_stats['std'], fmt='-o', color='b', ecolor='r',
             elinewidth=2, capsize=4, label='Avg Interest Rate')

# Customize the plot
plt.title('Average Interest Rate vs Loan Duration', fontsize=16)
plt.xlabel('Loan Duration (in months)', fontsize=12)
plt.ylabel('Average Interest Rate (%)', fontsize=12)
plt.legend()
plt.tight_layout()

# Show the plot
plt.show()
```

**key insights**

1.The longer the loan duration, the higher the average interest rate tends to be, though there is variability in rates for each loan duration, particularly for short-term and long-term loans.

2.This trend suggests that borrowers taking out loans for longer periods should expect higher interest rates, but they may experience some variability depending on the specific loan terms and market conditions.
