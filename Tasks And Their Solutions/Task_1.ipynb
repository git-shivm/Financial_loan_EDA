**1.Handling missing values using:
dropna() to remove missing values.
fillna() to fill missing values.**

**2.Handling duplicates: drop_duplicates()
Data type conversions using astype()**

```python
import pandas as pd
loan_df = pd.read_csv('/content/Loan.csv')
loan_df.info()
```

**Handling missing values using: dropna() to remove missing values. fillna() to fill missing values.**

```python
loan_df.isnull().sum()
#There is no null values in this dataset
```

Methods to drop nun values and fill other value in place of null

```python
loan_df.dropna(inplace=True) # Droping null values

loan_df.fillna(0, inplace=True) # Filling null values with 0
```

we can specify columns also

```python
loan_df.dropna(subset=['EducationLevel'], inplace=True)  # Droping null values from specific column
```

```python
loan_df.fillna({'EducationLevel': 'Unknown'}, inplace=True) # Filling null values from specific column
```

**Handling duplicates: drop_duplicates() Data type conversions using astype()**

```python
loan_df.duplicated().sum() # Checking duplicate values

loan_df.drop_duplicates(inplace=True) # Droping duplicate values
```

Data type conversion

here we converted object data type into datetime

```python
loan_df['ApplicationDate'] = pd.to_datetime(loan_df['ApplicationDate']) # Converting data type of column
```
