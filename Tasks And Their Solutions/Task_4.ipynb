# Total Loan Amount by Loan Purpose:


*  **What is the total loan amount for each loan purpose?**
* **Visualize the total loan amount for different loan purposes using a bar chart.**






```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
#some data clean up
loan_df['ApplicationDate'] = pd.to_datetime(loan_df['ApplicationDate'])
loan_df.info()
```

```python
# Calculate the average RiskScore for each HomeOwnershipStatus
home_ownership_risk = loan_df.groupby('LoanPurpose')['LoanAmount'].sum()
home_ownership_risk
```

```python

# Create the bar chart
plt.figure(figsize=(8, 6))

# Plot the bars
home_ownership_risk.plot(kind = "bar" , color=['blue', 'green', 'orange' , 'red' , "yellow"], width=0.5)

# Add labels and title
plt.xlabel('Loan Purpose', fontsize=12)
plt.ylabel ('Total Loan Amount in Billion (USD)', fontsize=12)
plt.title('Total Loan Amount by Loan Purpose', fontsize=14)

# Display the plot
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Key Insights:

As we can see, people whose loan purpose is home, they have demanded the maximum loan and also the maximum amount of loan has been given to them, whereas education and others have the least loan demand.
