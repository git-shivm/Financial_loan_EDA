***Box plot***

**Loan Amount by Education Level: This could show how loan amounts differ based on the borrower's education level.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
loan_df.head()

```

```python
# Generate a box plot of Loan Amount by Education Level with colored boxes and diamond markers for outliers
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=loan_df,
    x='EducationLevel',
    y='LoanAmount',
    palette='Set2',  # Set2 palette provides various colors for each category
    flierprops=dict(marker='D', color='red', markersize=6)  # 'D' for diamond shape
)
plt.title('Loan Amount by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Loan Amount')
plt.xticks(rotation=45)
plt.show()


```
