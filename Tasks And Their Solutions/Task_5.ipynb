***Age vs Loan Amount Analysis:***

**Create a scatter plot showing the relationship between Age and LoanAmount. Color the points based on the LoanApproved status**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
#
```

```python
loan_df = pd.read_csv('/content/Loan.csv')
#some data clean up
loan_df['ApplicationDate'] = pd.to_datetime(loan_df['ApplicationDate'])
loan_df.info()

```

```python

plt.figure(figsize=(10, 6))

# Plot the scatter plot
sns.scatterplot(data=loan_df, x='Age', y='LoanAmount', hue='LoanApproved', palette='coolwarm', alpha=0.7)

# Adding labels and title
plt.title('Age vs Loan Amount (Colored by Loan Approval Status)', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Loan Amount', fontsize=12)

# Display the legend and the plot
plt.legend(title='Loan Approved Status')
plt.show()
```
