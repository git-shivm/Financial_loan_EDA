 **plot a pie chart for EmploymentStatus showing the percentages of loan approval (1) and loan rejection (0)**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
loan_df = pd.read_csv("/content/Loan.csv")
loan_df.info()
```

```python
# Calculate the loan approval counts
loanapproved_count = loan_df.groupby(['LoanApproved'])['EmploymentStatus'].count()

# Plot a pie chart for loan approval (1) and loan rejection (0)
loanapproved_count.plot(kind='pie', autopct='%1.1f%%', labels=['Loan Rejected', 'Loan Approved'],
                        colors=['#ff9999','#66b3ff'], startangle=90, figsize=(6, 6))

plt.ylabel('')
plt.title('Loan Approval vs Rejection Based on Employment Status')
plt.legend(loc='upper right')
plt.show()

```
